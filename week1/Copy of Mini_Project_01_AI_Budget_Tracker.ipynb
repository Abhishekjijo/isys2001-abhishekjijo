## Part 3: Building the Budget Tracker

Use AI to help you write a program that:
- Allows users to input their monthly income
- Allows users to input up to 3 expenses  
- Calculates remaining budget
- Shows a warning if budget is negative

### 💡 Pro Tip: Be specific about requirements!

```python
# TASK: Write your detailed prompt for the budget tracker
"""
My Budget Tracker Prompt:
[I want to create a simple Python budget tracker program for beginners.
The program should ask the user to enter their monthly income and then enter 3 expenses.
After that, it should calculate how much money is left after the expenses are deducted from the income.
I also want the program to display the total expenses and the remaining budget in a clear and neat way.
If the user spends more money than their income, the program should display a warning message saying the budget is negative.
Please keep the code simple, readable, and easy to understand for someone learning Python.]
"""
```

### Paste and Test Your AI-Generated Budget Tracker

```python
# AI-GENERATED BUDGET TRACKER CODE

income = float(input("Enter your monthly income: "))

expense1 = float(input("Enter your first expense: "))
expense2 = float(input("Enter your second expense: "))
expense3 = float(input("Enter your third expense: "))

total_expenses = expense1 + expense2 + expense3
remaining_budget = income - total_expenses

print("\n------ Budget Summary ------")
print(f"Monthly Income: ${income:.2f}")
print(f"Total Expenses: ${total_expenses:.2f}")
print(f"Remaining Budget: ${remaining_budget:.2f}")

if remaining_budget < 0:
    print("⚠ Warning: Your budget is negative!")
else:
    print("✅ Your spending is within your budget.")
```

```python
# ENHANCED BUDGET TRACKER

income = float(input("Enter your monthly income: "))

expense1 = float(input("Enter your first expense: "))
expense2 = float(input("Enter your second expense: "))
expense3 = float(input("Enter your third expense: "))

total_expenses = expense1 + expense2 + expense3
remaining_budget = income - total_expenses

# Calculate percentage of income spent
percentage_spent = (total_expenses / income) * 100

print("\n------ Enhanced Budget Summary ------")
print(f"Monthly Income: ${income:.2f}")
print(f"Total Expenses: ${total_expenses:.2f}")
print(f"Remaining Budget: ${remaining_budget:.2f}")
print(f"Percentage of Income Spent: {percentage_spent:.2f}%")

if remaining_budget < 0:
    print("⚠️ Warning: Your budget is negative!")
else:
    print("✅ Your spending is within your budget.")
```
